# Explanation Stability Across Feature Sets
Loads results/explanation_stability.csv (produced by pipelines/run_all_experiments.py) and visualizes how consistent SHAP explanations are as the feature set shrinks from 49 to 15.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import matplotlib.pyplot as plt
from src.utils.config_loader import load_config, resolve_path

config = load_config("../configs/config.yaml")
stability_csv = resolve_path(config["experiments"]["stability_csv"])
df = pd.read_csv(stability_csv)
df
```

```python
df.plot(x="config_b", y=["rank_correlation", "cosine_similarity", "topk_overlap"], kind="bar", figsize=(9, 4))
plt.title("Explanation stability metrics across feature-set comparisons")
plt.tight_layout()
plt.show()
```
